# Exploratory Data Analysis (EDA) - Hasil Panen Sawit

Notebook ini bertujuan untuk mendemonstrasikan tahapan eksplorasi data secara visual sebelum data mentah dimasukkan ke dalam model *Deep Learning* LSTM. 

Analisis pola musiman (*seasonal trends*), distribusi fitur klimatologi, serta uji korelasi antar variabel disajikan di sini sebagai bentuk landasan teoretis penelitian Skripsi.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Set style
plt.style.use('ggplot')
sns.set_palette('viridis')
```

## 1. Memuat Dataset Gabungan

```python
# Load dataset from root folder
df = pd.read_csv('../dataset/Master_LSTM_Final_Large.csv')

# Hapus redundansi laporan
df = df.drop_duplicates()

print(f"Total Baris Bersih: {len(df)}")
display(df.head())
```

## 2. Analisis Distribusi Curah Hujan vs Yield

```python
fig, ax = plt.subplots(1, 2, figsize=(15, 6))

# Histogram Curah Hujan
sns.histplot(df['Rainfall_CH'], bins=30, kde=True, ax=ax[0], color='blue')
ax[0].set_title('Distribusi Curah Hujan Bulanan (mm)')
ax[0].set_xlabel('Curah Hujan (CH)')

# Histogram Target Yield
sns.histplot(df['Yield_Ton_Ha'], bins=30, kde=True, ax=ax[1], color='green')
ax[1].set_title('Distribusi Produktivitas (Ton/Ha)')
ax[1].set_xlabel('Yield (Ton/Ha)')

plt.tight_layout()
plt.show()
```

## 3. Matriks Korelasi

```python
plt.figure(figsize=(12, 10))
cols_to_correlate = ['Yield_Ton_Ha', 'Rainfall_CH', 'RainyDays_HH', 'Palm_Age', 
                     'CH_Lag_6', 'CH_Lag_12', 'CH_Lag_24']

correlation_matrix = df[cols_to_correlate].corr()

sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', fmt='.2f', linewidths=0.5)
plt.title('Heatmap Korelasi Pearson: Target vs Iklim & Lagging')
plt.show()
```
